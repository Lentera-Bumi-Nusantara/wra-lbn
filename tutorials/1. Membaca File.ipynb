## **1.1 File CSV**

#### **1.1.1. Sekilas tentang format csv**

Format *csv (comma separated values)* merupakan format untuk menyimpan tabel data yang sebagai teks yang data dari satu kolom dengan kolom lainnya dipisahkan oleh tanda koma `,` [1]. Format data ini biasanya menggunakan baris pertama sebagai nama kolom dan baris-baris berikutnya sebagai isi data untuk setiap kolom:

`kolom a, kolom b, kolom c, ..., kolom n`

`data a1, data b1, data c1, ..., data n1`

`   ... `

`data an, data bn, data cn, ..., data nn`

**Kelebihan & Kekurangan**

Format ini mudah dibaca menggunakan berbagai macam perangkat lunak, baik perangkat lunak *spreadsheet* (seperti Microsoft Excel, Open Office, Libre Office,) maupun perangkat lunak pembaca teks seperti notepad.

Karena data disimpan sebagai teks murni (`string`), ada beberapa kekurangan dari penyimpanan data menggunakan format ini:
- Perubahan data secara tidak disengaja. Hal ini bisa disebabkan oleh beberapa hal:
	- Perubahan data dari `string` menjadi `float` atau `integer` oleh pandas. Saat membaca kolom pada file csv, pandas biasanya akan membuat asumsi jenis data pada kolom tersebut. Misalnya data pada kolom dengan format '123.45' akan dianggap sebagai `float` secara otomatis, sedangkan data pada kolom dengan format '12345' akan dianggap sebagai `integer`.
	- Tidak sengaja mengubah data karena isi data terekspos secara langsung.  
- Ukuran file yang cukup besar

[1] https://en.wikipedia.org/wiki/Comma-separated_values

## **1.2. File Parquet**

#### **1.2.1 Sekilas tentang format Parquet**

Parquet merupakan format untuk menyimpan data tabulasi dalam sebagai data biner yang dikembangkan oleh Apache[1]. Setiap kolom pada parquet sudah disimpan sebagai tipe datanya masing-masing (`string`, `float`, `integer`, dll) secara terpisah dan dikompresi secara terpisah. 

**Kelebihan & Kekurangan**

Kelebihan menyimpan data dalam format parquet adalah ukuran file yang relatif kecil. Namun format ini biasanya tidak langsung dapat dibaca oleh perangkat lunak spreadsheet. Hal ini juga secara tidak langsung memberikan proteksi terhadap isi data dari modifikasi secara tidak sengaja. 

[1] https://en.wikipedia.org/wiki/Apache_Parquet 

#### **1.2.2 Membaca file parquet menggunakan pandas**

Perintah untuk membaca file csv menggunakan pandas adalah:

`df = pd.read_parquet(path_ke_file)`

Jalankan sel berikut untuk membaca file parquet:

````python
# Mendefinisikan path file
path = "LBN_2025_11.parquet"

# Membaca file dari path sebagai dataframe 'df':
df = pd.read_parquet(path)

# Melihat isi 5 baris pertama dari dataframe df: 
df
````

## **1.3. Membaca beberapa file secara bersamaan**

**PENTING!** Sebelum memulai bagian ini, siapkan sebuah folder berisi beberapa file csv atau parquet.

Untuk membaca beberapa file secara bersamaan, kita memerlukan bantuan library lainnya bernama `os`. Library ini berfungsi untuk membantu kita mengidentifikasi file-file yang ada di dalam suatu direktori. 

Untuk mengimpor library ini, jalankan sel berikut:

````python
import os
````

Dalam contoh ini, kita menyimpan beberapa data dari metmast di dalam folder `data` yang berada dalam satu folder yang sama dengan file ini. jalankan sel berikut untuk membaca file-file yang ada dalam folder data:

````python
# Mendefinisikan path:
path = "data"

# Mendapatkan daftar file di dalam path:
files = os.listdir(path)

# Lihat isi file di dalam direktori:
print(files)
````

Kita lihat bahwa ada beberapa file dalam format yang berbeda. Untuk bisa membaca beberapa file, kita perlu memilah hanya untuk file dengan format yang sama (hanya parquet atau hanya csv). Untuk melakukannya kita perlu menambahkan kondisi `endswith('format_yang_diinginkan')`

````python
# Mengambil daftar file parquet di dalam direktori:
files = [f for f in os.listdir(path) if f.endswith(".parquet")]

# Melihat daftar file yang sudah diseleksi:
print(files)
````

Selanjutnya gunakan loop `for` untuk membaca setiap file satu persatu dan menggabungkan file-file tersebut. Setiap file dapat dibaca menggunakan perintah `pd.read_csv()` atau `pd.read_parquet()` seperti pada bagian sebelumnya. Lalu digabungkan (`concatenate`) sebagai satu dataframe dimana pandas akan secara otomatis menggabungkan data dengan nama kolom yang sama. 

Syarat untuk bisa menggabungkan beberapa dataframe adalah dataframe-dataframe tersebut memiliki jumlah kolom yang sama. 

Jalankan sel berikut untuk menggabungkan file-file tersebut:

````python
# Siapkan list kosong untuk menyimpan sementara setiap dataframe dari setiap file parquet
dataframes = []

# Membaca setiap file di dalam variabel files:
for f in files:

	# Membaca file f sebagai dataframe df:
	df = pd.read_parquet(f)

	# Menambahkan dataframe 'df' ke dalam list 'dataframes':
	dataframes.append(df)

# Menggabungkan seluruh dataframe di dalam list dataframes sebagai satu dataframe:
dataframes = pd.concat(dataframes)

# Melihat isi dataframe
dataframes
````

#### **1.4.2 Membaca banyak file**

````python
# Mengimpor library
import pandas as pd 
import os 

# Mendefinisikan path berisi file-file data metmast:
path = "data"

# Mengambil daftar file parquet di dalam direktori:
files = [f for f in os.listdir(path) if f.endswith(".parquet")]

# Siapkan list kosong untuk menyimpan sementara setiap dataframe dari setiap file parquet
dataframes = []

# Membaca setiap file di dalam variabel files:
for f in files:

	# Membaca file f sebagai dataframe df:
	df = pd.read_parquet(f)

	# Menambahkan dataframe 'df' ke dalam list 'dataframes':
	dataframes.append(df)

# Menggabungkan seluruh dataframe di dalam list dataframes sebagai satu dataframe:
dataframes = pd.concat(dataframes)

# Melihat isi dataframe
dataframes
````
